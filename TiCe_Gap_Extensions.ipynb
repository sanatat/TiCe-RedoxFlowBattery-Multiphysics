# Ti–Ce Redox Flow Battery: Extended Research-Gap Models

Independent computational research prototype based on Fu et al. (2026), [DOI: 10.1002/cssc.202502732](https://doi.org/10.1002/cssc.202502732). **Unmeasured scenario parameters are not experiments or validated physical predictions.**

```python
from extensions.chemistry import MixedAcidHypothesis, mixed_acid_scenario
from extensions.reactor import ReactorScenario, simulate_discharge
from extensions.membrane import MembraneScenario, simulate_membrane
from extensions.stack import StackScenario, stack_operating_point
from extensions.lifetime import LifetimeScenario, simulate_lifetime
```

## 1. Conditional mixed-acid chemistry — NOT measured speciation or precipitation

The SI provides a **2 M MSA-only anchor**. Sulfate parameters below are hypothetical.

```python
for f in (0, .5, 1.):
    print(f, mixed_acid_scenario(MixedAcidHypothesis(sulfate_fraction=f))["conductivity_mS_cm_scenario"])
```

## 2. Axially coupled electrochemical reactor and transport

Uses SI geometric area and flow rate; effective film, surface and Darcy permeability are **not measured**.

```python
for j in (30, 50, 70, 90):
    r=simulate_discharge(ReactorScenario(j_mA_cm2=j))
    print(j, r["cell_voltage_V"], r["pressure_drop_Pa"], r["Ce_balance_relative_error"])
```

## 3. Membrane crossover (Ce SI permeability + hypothesized field and water transport)

```python
r=simulate_membrane(MembraneScenario(),hours=120,initial_concentrations_mol_m3=((500,0),(0,0)))
print(r["trace"][-1], r["final_Ce_mol"]-r["initial_Ce_mol"])
```

## 4. Stack network (series voltage, parallel Darcy flow), and 5. lifetime scenario

```python
print(stack_operating_point(StackScenario(n_cells=8))["net_power_W"])
life=simulate_lifetime(LifetimeScenario(cycles=500))
print("Remix events:", life["remix_cycles"], "Final normalized capacity:", life["final_capacity_fraction"])
```
